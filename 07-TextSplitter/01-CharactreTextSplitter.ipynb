```python
with open("./data/appendix-keywords.txt") as f:
    file = f.read()
```

```python
print(file[:500])
```

```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text_splitter = CharacterTextSplitter(
    separator="\n\n",  # 텍스트 분할 기본값은 "\n\n"
    chunk_size=210,  # 분할할 텍스트 청크의 최대 크기 (문자 수)
    chunk_overlap=0,  # 분할된 텍스트 청크 간의 중복되는 문자 수
    length_function=len  # 텍스트의 길이를 계산하는 함수
)
```

```python
texts = text_splitter.create_documents([file])
```

```python
print(len(texts[0].page_content))  # 분할된 문서의 개수
print(texts[0])  # 분할된 문서 중 첫 번째 문서
```

```python
# 메타데이터

metadatas = [
    {"document": 1},
    {"document": 2},
]
```

```python
documents = text_splitter.create_documents(
    [
        file, 
        file
    ], 
    metadatas=metadatas
)
```
