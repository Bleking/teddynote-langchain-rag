```python
from dotenv import load_dotenv

from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai.embeddings import OpenAIEmbeddings
```

```python
load_dotenv()
```

```python
embedding = OpenAIEmbeddings()
text_splitter = SemanticChunker(embedding)
```

텍스트 분할하기

```python
print(chunks[0])
```

```python
# 청크를 문서로 변환

docs = text_splitter.create_documents([file])
```

```python
print(docs[0].page_document)
```

# Breakpoints

문장을 분리할 시점 결정

percentile
- 백분위수 기반 분리 (기본적인 분리 방식)

```python
text_splitter1 = SemanticChunker(
    embedding, 
    breakpoint_threshold_type="percentile", 
    breakpoint_threshold_amount=70
)
```

```python
docs1 = text_splitter1.create_documents([file])

for i, doc1 in enumerate(docs1[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc1.page_content)  # 분할된 문서 중 첫 번째 문서의 내용
    print("===" * 20)
```

```python
print(len(docs1))
```

Standard deviation
- 지정한 breakpoint_threshold_amount 표준편차보다 큰 차이가 있으면 분할

```python
text_splitter2 = SemanticChunker(
    embedding, 
    breakpoint_threshold_type="standard_deviation", 
    breakpoint_threshold_amount=1.25
)
```

```python
docs2 = text_splitter2.create_documents([file])
```

```python
docs2 = text_splitter2.create_documents([file])
for i, doc2 in enumerate(docs2[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc2.page_content)  # 분할된 문서 중 첫 번째 문서의 내용을 출력합니다.
    print("===" * 20)
```

```python
print(len(docs2))
```

Interquartile
- 사분위수 범위

```python
text_splitter3 = SemanticChunker(
    embedding, 
    breakpoint_threshold_type="interquartile", 
    breakpoint_threshold_amount=0.5
)
```

```python
docs3 = text_splitter3.create_documents([file])
for i, doc3 in enumerate(docs3[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc3.page_content)  # 분할된 문서 중 첫 번째 문서의 내용을 출력합니다.
    print("===" * 20)
```

```python
print(len(docs3))
```
